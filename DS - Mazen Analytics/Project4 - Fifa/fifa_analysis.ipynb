## Concatenate FIFA Files

```python
df21 = pd.read_csv(r'data\FIFA21_official_data.csv')
df22 = pd.read_csv(r'data\FIFA22_official_data.csv')
df23 = pd.read_csv(r'data\FIFA23_official_data.csv')
```

```python
for i, col in enumerate(df21.columns):
  if col not in df23.columns:
    print(f"{i+1}- {col}")
```

### Units standardization (Kg, lbs, ', Cm)

```python
df = pd.read_excel("fifa_concatenated.xlsx", sheet_name='Working Sheet')
df.info()
```

```python
# Normalize 'Release Clause' column
df['Release Clause (Normalized)'] = ""
for i in range(df.shape[0]):
  if not pd.isnull(df.iloc[i]['Release Clause']):
    if df.iloc[i]['Release Clause'][-1] == "M":
      df.loc[i, 'Release Clause (Normalized)'] = float(df.iloc[i]['Release Clause'][1:-1]) * 1000000
    elif df.iloc[i]['Release Clause'][-1] == "K":
      df.loc[i, 'Release Clause (Normalized)'] = float(df.iloc[i]['Release Clause'][1:-1]) * 1000
    else:
      TypeError("Unknown format")
  else:
    df.loc[i, 'Release Clause (Normalized)'] = np.nan

# df['Release Clause (Normalized)'].astype('int')
```
